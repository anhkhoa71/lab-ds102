```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img_path = "zebra.jpg"
img = cv2.imread(img_path, cv2.IMREAD_COLOR)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
h, w, c = img.shape
plt.imshow(img)
plt.show()
```

```python
img_data = img.reshape((-1, 3))
img_data.shape
```

## Gaussian Mixture

```python
from sklearn.mixture import GaussianMixture
import warnings
warnings.filterwarnings("ignore")


gm = GaussianMixture(n_components=2, covariance_type="full", random_state=42)
gm.fit(img_data)
```

```python
labels = gm.predict(img_data)
labels = labels.reshape(h, w, -1)
plt.imshow(labels)
plt.show()
```

## Remove Background

```python
segmented_image = np.where(labels==0, 255, img).astype(np.uint8)
plt.imshow(segmented_image)
plt.show()
```
